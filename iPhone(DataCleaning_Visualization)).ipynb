## Clean Text Columns

```python
df["product_name"] = df["product_name"].str.strip()
```

```python
df["product_name"] = df["product_name"].str.lower()
```

```python
df["product_name"].str.len()
```

## Check Numeric Columns

```python
df[["sale_price","mrp","discount_percentage"]].dtypes
```

```python
df["sale_price"] = pd.to_numeric(df["sale_price"], errors="coerce")
df["mrp"] = pd.to_numeric(df["mrp"], errors="coerce")
df["discount_percentage"] = pd.to_numeric(
    df["discount_percentage"], errors="coerce"
)
```

## Handle Missing Values

```python
df.isnull().sum()
```

```python
df["sale_price"].median()
```

```python
df["sale_price"] = df["sale_price"].fillna(
    df["sale_price"].median()
)
```

```python
df["brand"] = df["brand"].fillna("Unknown")
```

```python
df.isnull().sum()
```

## Check Outliers

## Create New Column -- Actual Discount

```python
df["calculated_discount"] = (
    (df["mrp"] - df["sale_price"]) / df["mrp"]
) * 100
```

```python
df["calculated_discount"] = df["calculated_discount"].round(2)
```

```python
df[["sale_price","mrp","discount_percentage","calculated_discount"]].head()
```

## Compare Listed VS Calculated discount

```python
df["discount_difference"] = (
    df["discount_percentage"]
    - df["calculated_discount"]
)
```

```python
df[
[
    "discount_percentage",
    "calculated_discount",
    "discount_difference"
]
].head()
```

## Create Price Category

```python
df["price_category"] = pd.cut(
    df["sale_price"],
    bins=[0, 30000, 60000, 100000, np.inf],
    labels=["Budget", "Mid Range", "Premium", "Luxury"]
)
```

```python
df[["sale_price","price_category"]].head(10)
```

```python
df["price_category"].value_counts()
```

## Extract Storage

```python
df["storage_gb"] = df["product_name"].str.extract(
    r"(\d+)\s*GB",
    expand=False
)
```

```python
df["storage_gb"] = pd.to_numeric(
    df["storage_gb"],
    errors="coerce"
)
```

```python
df[["product_name", "storage_gb"]].head(10)
```

## Extract Ram

```python
df["ram_gb"] = df["ram"].str.extract(
    r"(\d+)",
    expand=False
)
```

```python
df["ram_gb"] = pd.to_numeric(
    df["ram_gb"],
    errors="coerce"
)
```

```python
df[["ram", "ram_gb"]].head()
```

## Aggregation

```python
df.groupby("brand").agg({
    "sale_price": ["mean", "min", "max"],
    "star_rating": "mean",
    "number_of_reviews": "sum"
})
```

```python
summary = df.groupby("brand").agg(
    product_count=("product_name", "count"),
    average_price=("sale_price", "mean"),
    minimum_price=("sale_price", "min"),
    maximum_price=("sale_price", "max"),
    average_rating=("star_rating", "mean")
)

summary
```

## Visualization

### Histogram

```python
plt.figure(figsize=(8,5))

plt.hist(df["sale_price"], bins=15)

plt.title("Distribution of Sale Price")
plt.xlabel("Sale Price")
plt.ylabel("Number of Products")

plt.show
```

### Seaborn Histogram

```python
sns.histplot(
    data=df,
    x="sale_price",
    bins=15,
    kde=True
)

plt.title("Distribution of iPhone Sale Prices")
plt.show()
```

### Boxplot

```python
sns.boxplot(
    data=df,
    y="sale_price"
)

plt.title("Sale Price Distribution")
plt.show()
```

### Bar Chart

```python
brand_price = df.groupby("brand")["sale_price"].mean()

brand_price.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Average Sale Price by Brand")
plt.xlabel("Brand")
plt.ylabel("Average Sale Price")

plt.show()
```

### Countplot

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="ram"
    
)

plt.title("Number of Product by RAM")
plt.xlabel("RAM")
plt.ylabel("Number of Products")

plt.show()
```

### Scatter plot

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="sale_price",
    y="star_rating"
)

plt.title("Sale Price vs Star Rating")
plt.xlabel("Sale Price")
plt.ylabel("Star Rating")

plt.show()
```

### Scatter Plot with RAM


```python
sns.scatterplot(
    data=df,
    x="sale_price",
    y="star_rating",
    hue="ram"
)

plt.title("Sale Price vs rating by RAM")
plt.show()
```

### Discount Distribution

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="discount_percentage",
    kde=True
)

plt.title("Discount Distribution")
plt.xlabel("Discount %")
plt.ylabel("Number of products")

plt.show()
```
